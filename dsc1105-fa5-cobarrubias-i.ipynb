## A. Data Preparation and Cleaning (5 points)


```r
# a. Import necessary libraries (1 pt)
# b. Load dataset correctly (1 pt)
library(tidyverse)
library(reshape2)

data <- read_csv("/kaggle/input/datasets/ivancobarrubias/adult-dataset1/adult1.csv")
```

```r
data <- as.tibble(data)
head(data)
```

```r
glimpse(data)
```

```r
# c. Assign column names if needed (1 pt)
# d. Recode education into 5 categories (1 pt)
# e. Handle missing values (1 pt)

data <- na.omit(data)
```

```r
data <- data %>%
  mutate(education_grouped = case_when(
    education %in% c("Preschool", "1st-4th", "5th-6th") ~ "Elementary",
    education %in% c("7th-8th", "9th") ~ "Middle School",
    education %in% c("10th", "11th", "12th", "HS-grad") ~ "High School",
    education %in% c("Some-college", "Assoc-acdm", "Assoc-voc") ~ "College / Associate",
    education %in% c("Bachelors", "Masters", "Doctorate", "Prof-school") ~ "Advanced",
    TRUE ~ "Other"
  ))
head(data)
```

## B. Contingency Table Construction (5 points)

```r
# a. Create contingency table using education_grouped vs income (2 pts)
# b. Add totals (1 pt)
contingency_table <- table(data$education_grouped, data$income)
ct_with_totals <- addmargins(contingency_table)

c. Compute row percentages (1 pt)
row_pct <- prop.table(contingency_table, 1) * 100

cat("Table 1: Contingency Table (Frequencies)\n")
print(ct_with_totals)
cat("\nRow Percentages:\n")
print(round(row_pct, 2))
```

## C. Chi-Square Test of Independence (5 points)|

```r
# a. Perform chi-square test (2 pts)
# b. Report:
# - Chi-square statistic
# - Degrees of freedom
# - p-value (2 pts)

chi_test <- chisq.test(contingency_table)


cat("Chi-square statistic:", chi_test$statistic, "\n")
cat("Degrees of freedom:", chi_test$parameter, "\n")
cat("p-value:", format.pval(chi_test$p.value), "\n")


```

c. State significance (1 pt)

The result is statistically significant (p < .05). Reject the null hypothesis.

## D. Expected Frequencies & Residuals (5 points)

```r
# a. Compute expected frequencies (2 pts)
expected_freq <- chi_test$expected

# b. Compute standardized residuals (2 pts)
std_residuals <- chi_test$residuals


# c. Present results in table form (1 pt)
cat("Expected Frequencies:\n")
print(round(expected_freq, 2))
cat("\nStandardized Residuals:\n")
print(round(std_residuals, 2))
```

## E. Data Visualization (5 points)


```r
res_melted <- melt(std_residuals)
colnames(res_melted) <- c("Education", "Income", "Residual")

ggplot(res_melted, aes(x = Income, y = Education, fill = Residual)) +
  geom_tile() +
  geom_text(aes(label = round(Residual, 2)), color = "black") +
  scale_fill_gradient2(low = "#E69F00", mid = "#FFFFFF", high = "#56B4E9", midpoint = 0) +
  theme_minimal() +
  labs(title = "Figure 1. Heatmap of Standardized Residuals for Education and Income",
       subtitle = "Visualizing deviations from expected values",
       caption = "Note: Standardized residuals |z| > 1.96 are significant at the .05 level.",
       x = "Income Category",
       y = "Education Level")
```
